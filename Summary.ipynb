# PHASE 6: BUSINESS INSIGHTS & RECOMMENDATIONS


**5 Key Insights from EDA**

1. **Smoking Status Has the Greatest Impact on Medical Costs**

The exploratory analysis revealed that smokers incur significantly higher hospital bills than non-smokers. The boxplots and scatter plots showed a clear difference in medical expenses between the two groups, indicating that smoking is one of the strongest drivers of healthcare costs.

2. **Medical Costs Increase with Age**

The scatter plot of Age vs Hospital Bill showed a positive relationship, indicating that older patients generally experience higher medical expenses due to increased healthcare utilization and age-related medical conditions.

3. **Higher BMI Is Associated with Higher Hospital Bills**

Patients with higher Body Mass Index (BMI) tended to have higher hospital bills. Individuals classified as overweight or obese incurred greater medical costs, suggesting that BMI is an important predictor of healthcare expenditure.

4. **Medical Costs Vary Across Nigerian States**

The regional analysis showed noticeable differences in average hospital bills among states. These variations may be due to differences in healthcare facilities, treatment costs, and access to medical services.

5. **Hospital Bill Distribution Is Positively Skewed**

The distribution of hospital bills was right-skewed, indicating that while most patients incurred moderate healthcare costs, a smaller number of patients generated exceptionally high medical expenses. These high-cost patients may require special attention in insurance pricing and risk management.

**3 insights from modeling**

1. **Tree-Based Models Outperformed Linear Regression**

Both the Random Forest Regressor and Gradient Boosting Regressor achieved better prediction performance than the Linear Regression model, demonstrating that medical costs are influenced by complex, non-linear relationships among patient characteristics.

2. **Smoking Status, BMI, and Age Were the Most Important Predictors**

Feature importance analysis identified Smoker Status, BMI, and Age as the most influential variables affecting hospital bill predictions. These variables contributed most to the model's predictive performance.

3. **The Best Model Produced Accurate and Reliable Predictions**

The model with the highest R² Score and the lowest MAE, MSE, and RMSE was selected as the final predictive model. Its performance indicates that machine learning can effectively estimate medical insurance costs and support insurance pricing decisions.

*Risk segmentation strategy*


To improve underwriting and pricing decisions, patients can be categorized into three risk groups based on their predicted medical costs and key risk factors.

# 1.  Low-Risk Patients

**Characteristics**:

Non-smokers

Young adults

Normal or healthy BMI

No or few children

Low predicted medical costs

**Strategy**:

Offer lower insurance premiums.

Encourage preventive healthcare through regular medical check-ups.

Provide wellness incentives to maintain healthy lifestyles.

# 2.  Medium-Risk Patients

**Characteristics**:

Middle-aged individuals

Slightly overweight BMI

Moderate predicted medical costs

May have one or more dependents

**Strategy**:

Apply moderate insurance premiums.

Recommend annual health screenings.

Offer health education and lifestyle improvement programs.

# 3.  High-Risk Patients

**Characteristics**:

Smokers

Older adults

High BMI (overweight or obese)

High predicted medical costs

**Strategy**:

Apply risk-based premium pricing.

Provide disease management and smoking cessation programs.

Encourage regular medical monitoring to reduce future healthcare costs.

# Policy pricing recommendations


Based on the analysis and model predictions, the following pricing strategies are recommended:

1. **Implement Risk-Based Premium Pricing**

Insurance premiums should reflect an individual's predicted medical risk. High-risk patients should pay higher premiums, while low-risk patients receive more affordable rates.

2. **Reward Healthy Behaviours**

Introduce premium discounts or incentives for non-smokers, individuals maintaining a healthy BMI, and customers who participate in preventive healthcare programs.

3. **Review Premiums Periodically**

Insurance premiums should be reassessed regularly based on updated health information, lifestyle changes, and claim history to ensure fair and accurate pricing.

# Suggestions for reducing high medical costs

1. **Promote Smoking Cessation Programs**

Since smokers were identified as having significantly higher medical costs, insurance providers should support smoking cessation initiatives through awareness campaigns, counseling, and incentive programs.

2. **Encourage Healthy Weight Management**

Implement wellness programs that promote balanced nutrition, regular physical activity, and weight management to reduce obesity-related health conditions and associated medical expenses.

3. **Increase Preventive Healthcare Services**

Encourage routine medical check-ups, vaccinations, and early disease screening to detect health conditions before they become severe and expensive to treat.

4. **Improve Health Education**

Provide educational resources on healthy lifestyle choices, chronic disease prevention, and personal health management to help reduce avoidable healthcare costs.

5. **Develop Data-Driven Insurance Policies**

Leverage predictive analytics and machine learning models to identify high-risk individuals early, enabling personalized interventions, efficient resource allocation, and more accurate insurance pricing.
